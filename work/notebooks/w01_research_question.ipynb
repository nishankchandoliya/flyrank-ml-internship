## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

I will use the starter CSV and treat one row as one pseudonymized content item. The file has 30,000 rows and 44 columns. In the starter data, **55.75% of pages (16,726 pages) have at least 500 impressions in the last 90 days**, so there is a substantial group of pages with measurable search visibility. Also, **32.48% of pages (9,745 pages) meet the starter's low-CTR visible-page condition**: at least 500 impressions, average position from 1 to 20, and CTR below 0.5%. These are concrete examples of pages where a review decision could matter.

The median number of 90-day impressions is **731 impressions per page**, which also shows that the starter set contains a mix of lower- and higher-visibility pages. These numbers make the lane worth investigating, but they do not prove that a refresh would improve performance.

```python
import pandas as pd

# Starter dataset: one row per pseudonymized content item
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

visible = df["impressions_90d"] >= 500
low_ctr_visible = (
    (df["impressions_90d"] >= 500)
    & (df["avg_position"].between(1, 20))
    & (df["ctr"] < 0.5)
)

print(f"Rows: {len(df):,}")
print(f"Clients: {df['client_id'].nunique():,}")
print(f"Pages with >=500 impressions_90d: {visible.sum():,} ({visible.mean()*100:.2f}%)")
print(f"Low-CTR visible pages: {low_ctr_visible.sum():,} ({low_ctr_visible.mean()*100:.2f}%)")
print(f"Median impressions_90d: {df['impressions_90d'].median():,.0f}")
```
